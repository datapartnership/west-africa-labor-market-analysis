## Software Development Trends using GitHub

### Developers

The number of developer accounts located in a given country based on self-reported location, as of the last date of each quarter.

**Assumptions & exclusions:**
- Excludes users that are bots.
- Excludes developers whose activity volume exceeded a threshold that could be reasonably attributed to human activity, and developers classified by automated systems as inauthentic — so the metric deliberately does not capture AI bots or automated accounts.
- Includes developer accounts that may no longer be active.
- Location is assigned by the mode of a developer's daily location (via IP address). If a developer contributes from the US for two months but also while traveling in India, all activity that quarter is assigned to the US. The last known location is also carried forward daily, even on days with no activity.

### Collaborators (Economy collaborators)

The volume of collaboration based on the sum of git pushes sent and pull requests opened by a developer to a repository owned by another developer or organization during each quarter.

**Assumptions & exclusions:**
- Location for repositories and organizations is assigned by the mode location of their members. Because of this, the metric may under-count cross-border collaboration, where contributions to a repo may involve users from many economies rather than just one.
- As such, economy collaborators should be viewed as a lower bound rather than a precise measure.

### Assumptions That Apply to All Metrics

- Only public activity is included; private repositories are excluded entirely.
- A metric is only reported for an economy when 100 or more unique developers performed that activity in the period, so smaller economies are missing.
- Activity is assigned to an economy by IP address, so VPNs and other IP-masking tools distort the metrics, affecting some economies more than others.
- GitHub data alone is acknowledged to paint an incomplete picture of software development — it's one resource in the ecosystem.

### New GitHub developer accounts in West Africa

GitHub's Innovation Graph shows an increase in the number of new developer accounts in West Africa over the period observed (2020 Q1 onward). While the growth rate of new developer accounts in Africa has been increasing steadily since 2019, it started to accelerate in 2024 Q4 and strengthened further in 2025 Q1. Across the West African countries, the quarterly increase rose from 35.9 to 42.4 per 100,000 working-age people between 2024 Q3 and Q4, then to 65.1 in 2025 Q1. 2024 Q4 is best interpreted as the beginning of an acceleration rather than the single largest quarterly peak. This acceleration occurred around the time of [GitHub's announcement of Copilot Free](https://github.blog/news-insights/product-news/github-copilot-in-vscode-free/) on December 18, 2024. The timing is consistent with a possible association between wider access to an AI coding assistant and the acceleration.

## West Africa GitHub Collaboration Inflows

Collaboration flowing into West African repositories has increased over the period, while also becoming more geographically diverse. In 2026 Q1, Europe & Central Asia remained the largest source region, followed by North America, with East Africa also making a meaningful contribution. Collaboration within West Africa was higher than at the beginning of the series, although it fluctuated and remained below its 2023 peak. Contributions from Central Africa have also become more visible in recent years, but they remain small and volatile. Overall, the pattern suggests that West African repositories are attracting participation from a broader set of regions, rather than becoming dependent on a single external region.

## West Africa GitHub Collaboration Outflows

Collaboration flowing from West African developers to repositories elsewhere has grown strongly. North America remained the largest destination in 2026 Q1, but its share of outflows declined from about 52 percent in 2020 Q1 to 35 percent, while the share going to Europe & Central Asia increased from about 25 percent to 33 percent. This narrowing gap indicates diversification away from the earlier concentration in North America, rather than its replacement as a major collaboration. Internal West African collaboration was also higher than at the beginning of the series but continued to fluctuate, while collaboration directed toward Central Africa rose from previously sparse reported levels and reached about 1.5 percent of outflows in 2026 Q1.
